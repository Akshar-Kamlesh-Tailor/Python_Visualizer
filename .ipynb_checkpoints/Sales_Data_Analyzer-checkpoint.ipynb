## 1. Environment Setup

```python
# Install if needed (uncomment)
# !pip install pandas matplotlib seaborn numpy

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["figure.dpi"] = 100

print("Libraries imported successfully!")
print(f"Pandas: {pd.__version__}")
print(f"NumPy:  {np.__version__}")
```

## 2. SalesDataAnalyzer Class (OOP)

```python
class SalesDataAnalyzer:
    """
    Encapsulates all sales data analysis and visualization functionality.
    Demonstrates Object-Oriented Programming principles.
    """

    def __init__(self, file_path=None):
        """Constructor: Initializes the analyzer and optionally loads data."""
        self.data= None
        self.file_path = file_path
        self.last_figure = None

        if file_path:
            self.load_data(file_path)

        print("SalesDataAnalyzer initialized successfully.")

    def __del__(self):
        """Destructor: Performs cleanup."""
        self.data = None
        self.last_figure = None

    # ------------------------------------------------------------------
    # Data Loading
    # ------------------------------------------------------------------
    def load_data(self, file_path):
        """Load data from a CSV file. Handles exceptions."""
        try:
            self.data = pd.read_csv(file_path)
            self.file_path = file_path

            if "date" in self.data.columns:
                self.data["date"] = pd.to_datetime(self.data["date"], errors="coerce")

            print(f"Dataset loaded successfully from '{file_path}'!")
            print(f"Shape: {self.data.shape[0]} rows × {self.data.shape[1]} columns")
            return True

        except FileNotFoundError:
            print(f"Error: File '{file_path}' does not exist.")
            return False
        except Exception as e:
            print(f"Error while loading data: {e}")
            return False
        """Load data from a CSV file. Handles exceptions."""
        try:

            self.data = pd.read_csv(file_path)
            self.file_path = file_path

            if "date" in self.data.columns:
                self.data["date"] = pd.to_datetime(self.data["date"], errors="coerce")

            print(f"Dataset loaded successfully from '{file_path}'!")
            print(f"Shape: {self.data.shape[0]} rows × {self.data.shape[1]} columns")
            return True

        except Exception as e:
            print(f"Error while loading data: {e}")
            return False

    # ------------------------------------------------------------------
    # Data Exploration
    # ------------------------------------------------------------------
    def explore_data(self) -> None:
        """Display basic information about the dataset."""
        if not self._check_data():
            return

        print("=" * 60)
        print("DATASET EXPLORATION")
        print("=" * 60)
        print("\n1. First 5 rows:")
        display(self.data.head())
        print("\n2. Last 5 rows:")
        display(self.data.tail())
        print("\n3. Column names:", list(self.data.columns))
        print("\n4. Data types:")
        print(self.data.dtypes)
        print("\n5. Info:")
        self.data.info()
        print("\n6. Shape:", self.data.shape)

    def show_head(self, n: int = 5):
        if self._check_data():
            display(self.data.head(n))

    def show_tail(self, n: int = 5):
        if self._check_data():
            display(self.data.tail(n))

    # ------------------------------------------------------------------
    # Data Cleaning
    # ------------------------------------------------------------------
    def clean_data(self) -> None:
        """Handle missing values and perform basic cleaning."""
        if not self._check_data():
            return

        print("=" * 60)
        print("DATA CLEANING")
        print("=" * 60)

        missing_before = self.data.isnull().sum().sum()
        print(f"\nTotal missing values before cleaning: {missing_before}")

        if missing_before == 0:
            print("No missing values found. Dataset is clean!")
            return

        print("\nMissing values per column:")
        print(self.data.isnull().sum()[self.data.isnull().sum() > 0])

        for col in self.data.columns:
            if self.data[col].isnull().sum() > 0:
                if pd.api.types.is_numeric_dtype(self.data[col]):
                    median_val = self.data[col].median()
                    self.data[col].fillna(median_val, inplace=True)
                    print(f"Filled '{col}' with median ({median_val})")
                else:
                    mode_val = self.data[col].mode()
                    if not mode_val.empty:
                        self.data[col].fillna(mode_val[0], inplace=True)
                        print(f"Filled '{col}' with mode ({mode_val[0]})")

        self.data.dropna(inplace=True)
        print(f"\nTotal missing values after cleaning: {self.data.isnull().sum().sum()}")

    # ------------------------------------------------------------------
    # Mathematical Operations + NumPy
    # ------------------------------------------------------------------
    def mathematical_operations(self) -> None:
        """Demonstrate NumPy array creation, indexing, slicing & element-wise ops."""
        if not self._check_data():
            return

        print("=" * 60)
        print("MATHEMATICAL OPERATIONS (with NumPy)")
        print("=" * 60)

        sale_arr = self.data["sale"].to_numpy()
        profit_arr = self.data["profit"].to_numpy()
        qty_arr = self.data["quantity"].to_numpy()

        print("\n--- Sale Array (NumPy) ---")
        print(f"Shape: {sale_arr.shape}")
        print(f"First 10 values: {sale_arr[:10]}")
        print(f"Slicing [50:60]: {sale_arr[50:60]}")

        print("\n--- Element-wise Operations ---")
        print(f"Sale * 1.1 (sample): {(sale_arr[:5] * 1.1).round(2)}")
        print(f"Log of sale (sample): {np.log1p(sale_arr[:5]).round(3)}")

        profit_margin = (profit_arr / sale_arr * 100).round(2)
        print(f"\nProfit margin % (first 10): {profit_margin[:10]}")
        print(f"Average profit margin: {np.mean(profit_margin):.2f}%")

        print("\n--- NumPy Statistical Summary on 'sale' ---")
        print(f"Sum:     {np.sum(sale_arr):,.2f}")
        print(f"Mean:    {np.mean(sale_arr):,.2f}")
        print(f"Std:     {np.std(sale_arr):,.2f}")
        print(f"Min:     {np.min(sale_arr):,.2f}")
        print(f"Max:     {np.max(sale_arr):,.2f}")
        print(f"25th %:  {np.percentile(sale_arr, 25):,.2f}")
        print(f"75th %:  {np.percentile(sale_arr, 75):,.2f}")

    # ------------------------------------------------------------------
    # Combine & Split
    # ------------------------------------------------------------------
    def combine_data(self, other_df: pd.DataFrame, how: str = "concat") -> None:
        """Combine current DataFrame with another."""
        if not self._check_data():
            return
        if how == "concat":
            self.data = pd.concat([self.data, other_df], ignore_index=True)
        elif how == "merge":
            self.data = pd.merge(self.data, other_df, how="outer")
        print(f"Combined. New shape: {self.data.shape}")

    def split_data(self, column: str = "region") -> dict:
        """Split DataFrame by unique values of a column."""
        if not self._check_data():
            return {}
        if column not in self.data.columns:
            print(f"Column '{column}' not found.")
            return {}
        groups = {val: grp.copy() for val, grp in self.data.groupby(column)}
        for val, grp in groups.items():
            print(f"Group '{val}': {len(grp)} rows")
        print(f"Total groups: {len(groups)}")
        return groups

    # ------------------------------------------------------------------
    # Search / Sort / Filter
    # ------------------------------------------------------------------
    def search_by_product(self, product: str):
        if not self._check_data():
            return
        result = self.data[self.data["product"].str.contains(product, case=False, na=False)]
        print(f"Found {len(result)} rows:")
        display(result.head(15))

    def filter_by_region(self, region: str):
        if not self._check_data():
            return
        result = self.data[self.data["region"].str.lower() == region.lower()]
        print(f"Found {len(result)} rows for '{region}':")
        display(result.head(15))

    def filter_by_min_sale(self, min_sale: float):
        if not self._check_data():
            return
        result = self.data[self.data["sale"] >= min_sale]
        print(f"Found {len(result)} rows with sale >= {min_sale}:")
        display(result.head(15))

    def sort_by(self, column: str, ascending: bool = False):
        if not self._check_data():
            return
        display(self.data.sort_values(by=column, ascending=ascending).head(15))

    # ------------------------------------------------------------------
    # Aggregations
    # ------------------------------------------------------------------
    def aggregate_functions(self) -> None:
        if not self._check_data():
            return
        print("=" * 60)
        print("AGGREGATING FUNCTIONS")
        print("=" * 60)

        numeric = self.data.select_dtypes(include=[np.number])
        print("\n--- Overall ---")
        display(numeric.agg(["sum", "mean", "count", "min", "max"]).round(2))

        print("\n--- Sales by Region ---")
        display(self.data.groupby("region")["sale"].agg(["sum", "mean", "count"]).round(2))

        print("\n--- Profit by Product ---")
        display(self.data.groupby("product")["profit"].agg(["sum", "mean", "count"]).round(2))

    # ------------------------------------------------------------------
    # Statistical Analysis
    # ------------------------------------------------------------------
    def statistical_analysis(self) -> None:
        if not self._check_data():
            return
        print("=" * 60)
        print("STATISTICAL ANALYSIS")
        print("=" * 60)
        print("\n--- describe() ---")
        display(self.data.describe().round(2))

        for col in ["sale", "profit", "quantity"]:
            if col in self.data.columns:
                arr = self.data[col].dropna().to_numpy()
                print(f"\n--- {col.upper()} ---")
                print(f"  Mean:     {np.mean(arr):,.2f}")
                print(f"  Std:      {np.std(arr, ddof=1):,.2f}")
                print(f"  Variance: {np.var(arr, ddof=1):,.2f}")
                print(f"  25%:      {np.percentile(arr, 25):,.2f}")
                print(f"  75%:      {np.percentile(arr, 75):,.2f}")
                print(f"  Skew:     {pd.Series(arr).skew():.3f}")

    # ------------------------------------------------------------------
    # Pivot Tables
    # ------------------------------------------------------------------
    def create_pivot_table(self) -> None:
        if not self._check_data():
            return
        print("=" * 60)
        print("PIVOT TABLES")
        print("=" * 60)

        print("\n--- Total Sales by Region × Product ---")
        pivot1 = pd.pivot_table(
            self.data, values="sale", index="region",
            columns="product", aggfunc="sum", fill_value=0
        )
        display(pivot1.round(2))

        print("\n--- Average Profit by Region × Category ---")
        pivot2 = pd.pivot_table(
            self.data, values="profit", index="region",
            columns="category", aggfunc="mean", fill_value=0
        )
        display(pivot2.round(2))

    # ------------------------------------------------------------------
    # Visualizations
    # ------------------------------------------------------------------
    def plot_bar(self):
        fig, ax = plt.subplots()
        self.data.groupby("region")["sale"].sum().sort_values(ascending=False).plot(
            kind="bar", color="steelblue", ax=ax
        )
        ax.set_title("Total Sales by Region", fontsize=14, fontweight="bold")
        ax.set_xlabel("Region")
        ax.set_ylabel("Total Sales")
        plt.xticks(rotation=45)
        plt.tight_layout()
        self.last_figure = fig
        plt.show()

    def plot_line(self):
        fig, ax = plt.subplots()
        monthly = self.data.set_index("date").resample("M")["sale"].sum()
        monthly.plot(kind="line", marker="o", color="darkorange", ax=ax)
        ax.set_title("Monthly Sales Trend", fontsize=14, fontweight="bold")
        ax.set_xlabel("Date")
        ax.set_ylabel("Total Sales")
        plt.tight_layout()
        self.last_figure = fig
        plt.show()

    def plot_scatter(self):
        fig, ax = plt.subplots()
        ax.scatter(self.data["sale"], self.data["profit"], alpha=0.6, c="teal", edgecolors="w")
        ax.set_title("Sale vs Profit", fontsize=14, fontweight="bold")
        ax.set_xlabel("Sale Amount")
        ax.set_ylabel("Profit")
        plt.tight_layout()
        self.last_figure = fig
        plt.show()

    def plot_pie(self):
        fig, ax = plt.subplots()
        sales_by_cat = self.data.groupby("category")["sale"].sum()
        ax.pie(sales_by_cat, labels=sales_by_cat.index, autopct="%1.1f%%",
               startangle=90, colors=sns.color_palette("pastel"))
        ax.set_title("Sales Distribution by Category", fontsize=14, fontweight="bold")
        plt.tight_layout()
        self.last_figure = fig
        plt.show()

    def plot_histogram(self):
        fig, ax = plt.subplots()
        ax.hist(self.data["sale"], bins=30, color="mediumpurple", edgecolor="black", alpha=0.8)
        ax.set_title("Distribution of Sale Amounts", fontsize=14, fontweight="bold")
        ax.set_xlabel("Sale Amount")
        ax.set_ylabel("Frequency")
        plt.tight_layout()
        self.last_figure = fig
        plt.show()

    def plot_stacked(self):
        pivot = self.data.pivot_table(values="sale", index="region",
                                      columns="category", aggfunc="sum", fill_value=0)
        fig, ax = plt.subplots()
        pivot.plot(kind="bar", stacked=True, ax=ax, colormap="Set2")
        ax.set_title("Stacked Sales by Region & Category", fontsize=14, fontweight="bold")
        ax.set_xlabel("Region")
        ax.set_ylabel("Total Sales")
        ax.legend(title="Category", bbox_to_anchor=(1.05, 1), loc="upper left")
        plt.xticks(rotation=45)
        plt.tight_layout()
        self.last_figure = fig
        plt.show()

    def plot_boxplot(self):
        fig, ax = plt.subplots(figsize=(11, 6))
        sns.boxplot(data=self.data, x="region", y="profit", palette="Set3", ax=ax)
        ax.set_title("Profit Distribution by Region (Seaborn)", fontsize=14, fontweight="bold")
        plt.xticks(rotation=45)
        plt.tight_layout()
        self.last_figure = fig
        plt.show()

    def plot_heatmap(self):
        numeric = self.data.select_dtypes(include=[np.number])
        corr = numeric.corr()
        fig, ax = plt.subplots(figsize=(8, 6))
        sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5, ax=ax)
        ax.set_title("Correlation Heatmap", fontsize=14, fontweight="bold")
        plt.tight_layout()
        self.last_figure = fig
        plt.show()

    def plot_dashboard(self):
        fig, axes = plt.subplots(2, 2, figsize=(14, 10))
        fig.suptitle("Sales Data Dashboard", fontsize=16, fontweight="bold")

        self.data.groupby("region")["sale"].sum().sort_values().plot(
            kind="barh", ax=axes[0, 0], color="steelblue")
        axes[0, 0].set_title("Sales by Region")

        monthly = self.data.set_index("date").resample("M")["sale"].sum()
        monthly.plot(ax=axes[0, 1], color="darkorange", marker="o")
        axes[0, 1].set_title("Monthly Sales Trend")

        cat_sales = self.data.groupby("category")["sale"].sum()
        axes[1, 0].pie(cat_sales, labels=cat_sales.index, autopct="%1.1f%%", startangle=90)
        axes[1, 0].set_title("Sales by Category")

        axes[1, 1].hist(self.data["profit"], bins=25, color="seagreen",
                        edgecolor="black", alpha=0.8)
        axes[1, 1].set_title("Profit Distribution")

        plt.tight_layout(rect=[0, 0, 1, 0.96])
        self.last_figure = fig
        plt.show()

    def save_visualization(self, filename: str = "plot.png"):
        if self.last_figure is None:
            print("No figure available. Generate a plot first.")
            return
        self.last_figure.savefig(filename, dpi=150, bbox_inches="tight")
        print(f"Saved as '{filename}'")

    # ------------------------------------------------------------------
    # Helpers
    # ------------------------------------------------------------------
    def _check_data(self) -> bool:
        if self.data is None or self.data.empty:
            print("No data loaded. Please load a dataset first.")
            return False
        return True

    def groupby_transform_demo(self):
        if not self._check_data():
            return
        self.data["region_avg_sale"] = self.data.groupby("region")["sale"].transform("mean")
        self.data["sale_vs_region_avg"] = self.data["sale"] - self.data["region_avg_sale"]
        print("Added: region_avg_sale, sale_vs_region_avg")
        display(self.data[["region", "sale", "region_avg_sale", "sale_vs_region_avg"]].head(10).round(2))


# Inheritance example
class AdvancedSalesAnalyzer(SalesDataAnalyzer):
    """Demonstrates inheritance."""

    def __init__(self, file_path=None):
        super().__init__(file_path)
        print("AdvancedSalesAnalyzer ready.")

    def top_n_products(self, n: int = 5):
        if not self._check_data():
            return
        top = self.data.groupby("product")["sale"].sum().nlargest(n)
        print(f"Top {n} products by total sales:")
        display(top.round(2))

print("Classes defined successfully!")
```
